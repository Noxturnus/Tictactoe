```python
class Player():
    def __init__(self, name):
        self.name = name
        self.wins = 0

class Game_Tic_Tac_Toe():
    def __init__(self):
        self.rows = 3
        self.columns = 3
        self.empty_cell = " "
        self.o = "O"
        self.x = "X"

        self.player_one = Player(input("Enter player one's name: "))
        self.player_two = Player(input("Enter player two's name: "))
        self.total_games = -1
        
    def tie(self):
        tie_rows = []
        for row in self.board:
            tie_rows.append('|'.join(row))
        print('\n' + '\n------\n'.join(tie_rows)+'\n')

    def show_score(self):
        print(f"\nTotal Games: {self.total_games}\n"
              f"{self.player_one.name}: {self.player_one.wins}\n"
              f"{self.player_two.name}: {self.player_two.wins}\n"
              f"Ties: {self.total_games - self.player_one.wins - self.player_two.wins}\n")
        
    def initialize_game_state(self):
        self.board = [[self.empty_cell for _ in range(self.columns)] for _ in range(self.rows)]
        self.current_turn = 0
        self.total_games += 1

    def check_winner(self):
        results = {"winner_found": False, "player": None}

        for row in self.board:
            if row[0] != self.empty_cell and row[0] == row[1] == row[2]:
                results["winner_found"] = True
                results["player"] = self.player_o if row[0] == self.o else self.player_x
                return results

        for col in range(self.columns):
            if self.board[0][col] != self.empty_cell and self.board[0][col] == self.board[1][col] == self.board[2][col]:
                results["winner_found"] = True
                results["player"] = self.player_o if self.board[0][col] == self.o else self.player_x
                return results

        if self.board[0][0] != self.empty_cell and self.board[0][0] == self.board[1][1] == self.board[2][2]:
            results["winner_found"] = True
            results["player"] = self.player_o if self.board[0][0] == self.o else self.player_x
            return results

        if self.board[2][0] != self.empty_cell and self.board[2][0] == self.board[1][1] == self.board[0][2]:
            results["winner_found"] = True
            results["player"] = self.player_o if self.board[2][0] == self.o else self.player_x
            return results

        return results

    def playable_rows(self):
        rows = []
        for row_num, row in enumerate(self.board):
            if self.empty_cell in row:
                rows.append(row_num)
        return rows
        
    def playable_columns(self, row):
        columns = []
        for column_num, column in enumerate(self.board[row]):
            if column == self.empty_cell:
                columns.append(column_num)
        return columns

    def play(self):
        self.initialize_game_state()
        self.show_score()

        if self.total_games % 2 == 0:
            self.player_o = self.player_one
            self.player_x = self.player_two
        else:
            self.player_o = self.player_two
            self.player_x = self.player_one

        winner_check = {"winner_found": False}

        while self.current_turn < 9:
            current_player = self.player_o if self.current_turn % 2 == 0 else self.player_x
            self.tie()
            print(f"It is {current_player.name}'s turn!")
            
            playable_rows_list = self.playable_rows()
            chosen_row = -1
            while chosen_row not in playable_rows_list:
                try:
                    row_input = input(f"Select a row - {playable_rows_list}: ")
                    chosen_row = int(row_input)
                except:
                    pass

            playable_cols_list = self.playable_columns(chosen_row)
            chosen_column = -1
            while chosen_column not in playable_cols_list:
                try:
                    column_input = input(f"Choose a column - {playable_cols_list}: ")
                    chosen_column = int(column_input)
                except:
                    pass

            self.board[chosen_row][chosen_column] = self.o if self.current_turn % 2 == 0 else self.x
            
            winner_check = self.check_winner()
            if winner_check["winner_found"]:
                self.tie()
                winner_check['player'].wins += 1
                print(f"{winner_check['player'].name} is the winner!")
                break
            
            self.current_turn += 1

        if not winner_check["winner_found"] and self.current_turn == 9:
            self.tie()
            print("The game has ended in a tie!")

        return input("Do you want to play again? (Y/y)") in ['Y', 'y']

t3 = Game_Tic_Tac_Toe()
play_again = True
while play_again:
    play_again = t3.play()
```
